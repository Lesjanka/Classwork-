```python
import asyncio
import random

score = 0

async def timer_task(seconds: int):
  for i in range(seconds, 0, -1):
    print(f"\rОсталось: {i} сек. | Решено: {score}", end="", flush=True)
    await asyncio.sleep(1)
  print("ВРЕМЯ ВЫШЛО! Игра окончена")

async def quiz_task():
  global score
  while True:
    a = random.randint(2, 20)
    b = random.randint(2, 20)
    op = random.choice(["+", "-", "*"])

    if op == "+":
      correct = a + b
    elif op == "-":
      correct = a - b
    else:
      correct = a * b

    question = f"{a} {op} {b} = ?"

    answer = await asyncio.to_thread(input, f"\n{question}")

    if int(answer) == correct:
      score += 1
      print("Верно!")
    else:
      print(f"Ошибка. Правильный ответ: {correct}")


async def main():
  print("МАТЕМАТИЧЕСКИЙ СПРИНТ! У тебя 60 секунд")
  timer = asyncio.create_task(timer_task(60))
  quiz = asyncio.create_task(quiz-task())

  done, pending = await asyncio.wait(
      [timer,quiz],
      return_when=asyncio.FIRST_COMPLETED
  )

  for task in pending:
    task.cancel()
    try:
      await task
    except asyncio.CancelledError:
      pass

    print(f"\nИТОГ: Ты решил {score} примеров за минуту!")

if __name__ == "__main__":
  asyncio.run(main())
```
